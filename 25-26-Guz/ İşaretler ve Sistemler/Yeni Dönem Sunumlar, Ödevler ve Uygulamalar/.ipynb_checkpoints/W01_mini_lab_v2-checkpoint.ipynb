
# W01 Mini-Lab — Sinyal Temelleri (CT vs DT), Dönüşümler, Tek/Çift Ayrımı

**Hedefler**
- Sürekli-zaman (CT) ve ayrık-zaman (DT) sinyal kavramlarını sezgisel olarak gözlemlemek
- Zaman kaydırma (ileri/geri) ve **genlik/zaman ölçekleme** işlemlerini uygulamak
- Tek/çift bileşen ayrımı yapmak ve doğrulamak
- Ses ve **PPG-benzeri** örneklerle kısa uygulamalar yapmak

```python

import numpy as np
import matplotlib.pyplot as plt

def newfig():
    plt.figure()
```


## A) CT vs DT — Basit "ses" örneği

Aşağıda 440 Hz'lik bir sinüs kısa bir süre için önce "sürekli-zaman benzeri" (çok yüksek örneklemeli)
daha sonra farklı örnekleme frekansları ile yeniden örneklenir.


```python

# "CT-benzeri" temsil
f = 440.0
T = 0.02
fs_ct = 100_000
t_ct = np.linspace(0, T, int(T*fs_ct), endpoint=False)
x_ct = np.sin(2*np.pi*f*t_ct)

newfig()
plt.plot(t_ct, x_ct)
plt.title("CT-benzeri temsil (yüksek örnekleme) — 440 Hz sinüs")
plt.xlabel("Zaman (s)"); plt.ylabel("Genlik")
plt.grid(True); plt.show()
```

```python

# DT temsiller
for fs in [8000, 2000, 1200, 900]:
    n = np.arange(0, int(T*fs))
    t = n/fs
    x = np.sin(2*np.pi*f*t)
    newfig()
    plt.stem(t, x)  # use_line_collection kullanılmıyor
    plt.title(f"DT temsil — fs = {fs} Hz, f = {f} Hz")
    plt.xlabel("Zaman (s)"); plt.ylabel("Genlik")
    plt.grid(True); plt.show()
```


**Not:** f=440 Hz sabitken örnekleme frekansı Fs düştükçe örnekler arası zaman aralığı büyür ve dalga şekli seyrekleşir. Çok düşük Fs değerlerinde aliasing riski oluşur.



## B) PPG-benzeri sinyal ve temel işlemler

Basit bir **PPG-benzeri** dalga formu (nabız) için periyodik darbe + eksponans karışımı kullanıyoruz.
Üzerinde **zaman kaydırma** ve **genlik/zaman ölçekleme** işlemlerini gözlemleyeceğiz.


```python

# PPG-benzeri dalga
fs = 100.0
T = 6.0
t = np.arange(0, T, 1/fs)

hr_hz = 1.1
base = 0.1*np.sin(2*np.pi*0.2*t)
ppg = np.zeros_like(t)
for k in range(int(T*hr_hz)+5):
    tc = k/hr_hz
    ppg += np.exp(-0.5*((t-tc)/(0.05))**2)
    ppg += 0.2*np.exp(-(t-tc-0.03)/0.2)*(t>tc)

x = 0.3 + base + 0.7*ppg

# Orijinal
newfig(); plt.plot(t, x); plt.title("PPG-benzeri sinyal (orijinal)")
plt.xlabel("Zaman (s)"); plt.ylabel("Genlik"); plt.grid(True); plt.show()

# Zaman kaydırma (gecikme)
D = 0.4
x_delayed = np.interp(t-D, t, x, left=0, right=0)
newfig(); plt.plot(t, x, label="orijinal"); plt.plot(t, x_delayed, label="gecikmiş")
plt.title("Zaman kaydırma örneği"); plt.xlabel("Zaman (s)"); plt.ylabel("Genlik")
plt.legend(); plt.grid(True); plt.show()
```


### Genlik ölçekleme

$y(t)=a\,x(t)$ — genlik büyütme/küçültme.


```python

def amp_scale(sig, a):
    return a * sig

for a in [0.5, 1.0, 1.5]:
    newfig()
    plt.plot(t, x, label="orijinal")
    plt.plot(t, amp_scale(x, a), label=f"genlik a={a}")
    plt.title("Genlik ölçekleme örneği")
    plt.xlabel("Zaman (s)"); plt.ylabel("Genlik")
    plt.legend(); plt.grid(True); plt.show()
```


### Zaman ölçekleme (sıkıştırma / genişletme)

Zaman ölçekleme **y(t) = x(a\,t)** ile tanımlanır.  
- **a > 1** ⇒ **sıkıştırma** (periyot kısalır, olaylar hızlanır)  
- **0 < a < 1** ⇒ **genişletme** (periyot uzar, olaylar yavaşlar)


```python

def time_scale(sig, t, a):
    # y(t) = x(a t). t' = a t. t üzerinde örnek almak için interpolasyon.
    if a == 1:
        return sig
    t_prime = a * t
    return np.interp(t, t_prime, sig, left=0.0, right=0.0)

for a in [0.7, 1.0, 2.4]:
    newfig()
    plt.plot(t, x, label="orijinal")
    plt.plot(t, time_scale(x, t, a), label=f"zaman ölçek a={a}")
    plt.title("Zaman ölçekleme — PPG-benzeri sinyal")
    plt.xlabel("Zaman (s)"); plt.ylabel("Genlik")
    plt.legend(); plt.grid(True); plt.show()
```


## C) Tek/Çift Bileşen Ayrımı (DT)

Tanımlar (DT):
\begin{align}
x_e[n] &= \tfrac{1}{2}\big(x[n] + x[-n]\big) \\
x_o[n] &= \tfrac{1}{2}\big(x[n] - x[-n]\big)
\end{align}


```python

N = 10
n = np.arange(-N, N+1)
xseq = np.sin(0.4*np.pi*n) + 0.3*(n==0)

def x_of(idx):
    # dışındaki değerlere 0 uzatma
    if -N <= idx <= N:
        return float(xseq[idx+N])
    return 0.0

xe = np.array([0.5*(x_of(int(ni)) + x_of(int(-ni))) for ni in n])
xo = np.array([0.5*(x_of(int(ni)) - x_of(int(-ni))) for ni in n])

newfig(); plt.stem(n, xseq); plt.title("Orijinal dizi x[n]")
plt.xlabel("n"); plt.ylabel("Genlik"); plt.grid(True); plt.show()

newfig(); plt.stem(n, xe); plt.title("Çift bileşen x_e[n]")
plt.xlabel("n"); plt.ylabel("Genlik"); plt.grid(True); plt.show()

newfig(); plt.stem(n, xo); plt.title("Tek bileşen x_o[n]")
plt.xlabel("n"); plt.ylabel("Genlik"); plt.grid(True); plt.show()

newfig(); plt.stem(n, xseq - (xe+xo)); plt.title("Doğrulama: x - (x_e + x_o)")
plt.xlabel("n"); plt.ylabel("Hata"); plt.grid(True); plt.show()
```
