# W02 Mini‑Lab: Birim Basamak, Birim Dürtü, Üstel ve Sinüzoidal Sinyaller

Bu çalışmada aşağıdaki konular yer alır:
- Sürekli zamanda u(t), e^{-a t}u(t) ve A·cos(2π f₀ t + φ) sinyalleri
- Ayrık zamanda u[n] ve δ[n]
- Zaman ölçekleme, kaydırma ve yansıma işlemleri (CT ve DT)
- Sayısal enerji ve güç hesapları (trapez kuralı ve pencere ortalaması)
- Basit bir fotopletismografi (PPG) sinyali örneği


```python
import numpy as np
import matplotlib.pyplot as plt

# Sürekli zaman ekseni (örneklenmiş)
t_min, t_max, t_N = -2.0, 2.0, 4001
t = np.linspace(t_min, t_max, t_N)
dt = t[1] - t[0]

# Ayrık zaman ekseni
n_min, n_max = -40, 40
n = np.arange(n_min, n_max+1)

# Sinyal parametreleri
a_exp = 1.0      # e^{-a t}u(t) için a>0
A_sin = 1.0      # sinüs genliği
f0_sin = 2.0     # sinüs frekansı (Hz)
phi_sin = 0.0    # sinüs fazı (radyan)

# Zaman işlemleri (CT)
t0_shift = 0.5   # sağa kaydırma (gecikme)
a_scale = 0.5    # ölçekleme: x(a·t); a<1 genleşme, a>1 sıkışma

# Zaman işlemleri (DT)
n0_shift = 3     # sağa kaydırma (örnek sayısı)
L_stretch = 2    # genleşme: y[n] = x[n//L]
M_compress = 2   # sıkıştırma: y[n] = x[n*M]

def u_ct(x):
    """Sürekli zamanda birim basamak."""
    return (x >= 0).astype(float)

def u_dt(k):
    """Ayrık zamanda birim basamak."""
    return (k >= 0).astype(float)

def delta_dt(k):
    """Ayrık zamanda birim dürtü: δ[k] = 1 (k = 0), aksi 0."""
    return (k == 0).astype(float)

def exp_ct(x, a=1.0):
    """Sönümlü üstel: e^{-a t}u(t)."""
    return np.exp(-a*np.maximum(x, 0.0)) * u_ct(x)

def sin_ct(x, A=1.0, f0=1.0, phi=0.0):
    """Sinüzoidal: A cos(2π f0 t + φ)."""
    return A*np.cos(2*np.pi*f0*x + phi)

def energy_trapz(x, dt):
    """Sürekli zaman enerji yaklaşığı: E ≈ ∫ |x(t)|^2 dt.
    Burada **np.trapz** (trapez kuralı) kullanılır: eşit aralıklı örneklerde alanı
    parça parça doğrusal yaklaşım ile hesaplar."""
    return np.trapz(np.abs(x)**2, dx=dt)

def power_window(x):
    """Ortalama güç yaklaşımı: P ≈ mean(|x|^2).
    Sonsuz zaman ortalaması yerine pratikte bir pencere üzerinden hesaplanır."""
    x = np.asarray(x)
    return np.mean(np.abs(x)**2)
```

```python
# Temel sürekli zaman sinyalleri
x_u   = u_ct(t)
x_exp = exp_ct(t, a=a_exp)
x_sin = sin_ct(t, A=A_sin, f0=f0_sin, phi=phi_sin)

plt.figure(); plt.plot(t, x_u);   plt.title('Birim Basamak u(t)');      plt.xlabel('t'); plt.ylabel('u(t)'); plt.grid(True)
plt.figure(); plt.plot(t, x_exp); plt.title('Üstel e^{-a t}u(t)');      plt.xlabel('t'); plt.ylabel('x(t)'); plt.grid(True)
plt.figure(); plt.plot(t, x_sin); plt.title('Sinüs A cos(2π f₀ t + φ)'); plt.xlabel('t'); plt.ylabel('x(t)'); plt.grid(True)
```

```python
# Ayrık zaman sinyalleri ve işlemleri
u_n = u_dt(n)
d_n = delta_dt(n)

def dt_shift(x, n, n0):
    """Kaydırma: y[n] = x[n - n0]."""
    y = np.zeros_like(x)
    for i, ni in enumerate(n):
        src = ni - n0
        if n_min <= src <= n_max:
            y[i] = x[src - n_min]
    return y

def dt_stretch(x, n, L):
    """Genleşme (tam sayı L): y[n] = x[n//L]."""
    L = int(L)
    y = np.zeros_like(x)
    for i, ni in enumerate(n):
        idx = ni//L if L>0 else 0
        y[i] = x[idx - n_min] if (L>0 and n_min<=idx<=n_max) else 0
    return y

def dt_compress(x, n, M):
    """Sıkıştırma (tam sayı M): y[n] = x[n*M]."""
    M = int(M)
    y = np.zeros_like(x)
    for i, ni in enumerate(n):
        src = ni * M
        if n_min <= src <= n_max:
            y[i] = x[src - n_min]
    return y

plt.figure(); plt.stem(n, u_n); plt.title('u[n]'); plt.xlabel('n'); plt.ylabel('u[n]'); plt.grid(True)
plt.figure(); plt.stem(n, d_n); plt.title('δ[n]'); plt.xlabel('n'); plt.ylabel('δ[n]'); plt.grid(True)
plt.figure(); plt.stem(n, dt_shift(u_n,n,n0_shift));     plt.title(f'u[n-n0], n0={n0_shift}'); plt.xlabel('n'); plt.ylabel('u[n-n0]'); plt.grid(True)
plt.figure(); plt.stem(n, dt_stretch(u_n,n,L_stretch));  plt.title(f'u[n//L], L={L_stretch} (genleşme)'); plt.xlabel('n'); plt.ylabel('u[n//L]'); plt.grid(True)
plt.figure(); plt.stem(n, dt_compress(u_n,n,M_compress));plt.title(f'u[n*M], M={M_compress} (sıkıştırma)'); plt.xlabel('n'); plt.ylabel('u[n*M]'); plt.grid(True)
```

```python
# Sürekli zamanda zaman işlemleri (u, üstel, sinüs)
def scale_sig(f, t, a): return f(a*t)
def shift_sig(f, t, t0): return f(t - t0)
def reflect_sig(f, t):   return f(-t)

sig_u   = lambda x: u_ct(x)
sig_exp = lambda x: exp_ct(x, a=a_exp)
sig_sin = lambda x: sin_ct(x, A=A_sin, f0=f0_sin, phi=phi_sin)

for name, f in [('u(t)',sig_u), ('e^{-a t}u(t)',sig_exp), ('A cos(2π f₀ t + φ)',sig_sin)]:
    plt.figure(); plt.plot(t, scale_sig(f,t,a_scale)); plt.title(f'Ölçekleme: {name}, a={a_scale}'); plt.xlabel('t'); plt.ylabel('x(t)'); plt.grid(True)
    plt.figure(); plt.plot(t, shift_sig(f,t,t0_shift)); plt.title(f'Kaydırma: {name}, t0={t0_shift}'); plt.xlabel('t'); plt.ylabel('x(t)'); plt.grid(True)
    plt.figure(); plt.plot(t, reflect_sig(f,t));        plt.title(f'Yansıma: {name}');               plt.xlabel('t'); plt.ylabel('x(t)'); plt.grid(True)
```

```python
# Enerji ve güç örnekleri (sayısal + teorik)
E_exp_num = energy_trapz(x_exp, dt)
E_exp_the = 1.0/(2.0*a_exp)
P_sin_win = power_window(x_sin)
P_sin_the = (A_sin**2)/2.0
print('Enerji sinyali  e^{-a t}u(t):  E_sayisal ≈', E_exp_num, '  E_teorik =', E_exp_the)
print('Güç   sinyali  A cos(2π f₀ t + φ):  P_pencere ≈', P_sin_win, '  P_teorik =', P_sin_the)

# Notlar:
# - e^{-a t}u(t) sönümlendiği için |x(t)|^2 integrali sonludur → enerji sinyali.
# - Sinüs periyodiktir; enerji integrali diverge eder. Ortalama güç (mean(|x|^2)) sonludur → güç sinyali.
```

```python
# PPG örneği: iki Gauss darbesi (sistolik tepe + diyastolik omuz) + baz çizgi + yavaş modülasyon
fs, T = 2000, 4.0
tt = np.arange(0, T, 1/fs)
HR_hz = 1.25  # ~75 bpm
beat_times = np.arange(0.4, T, 1/HR_hz)

def ppg_pulse(t, t0, amp1=0.9, w1=0.025, amp2=0.35, delay2=0.05, w2=0.015):
    g1 = amp1*np.exp(-((t - t0)/w1)**2)
    g2 = amp2*np.exp(-((t - (t0 + delay2))/w2)**2)
    return g1 + g2

ppg = np.zeros_like(tt) + 0.2
for t0 in beat_times:
    ppg += ppg_pulse(tt, t0)
ppg += 0.02*np.sin(2*np.pi*0.25*tt)

plt.figure(); plt.plot(tt[:2000], ppg[:2000]);
plt.title('PPG – ilk 1 s'); plt.xlabel('t [s]'); plt.ylabel('PPG'); plt.grid(True)
win = ppg[:2000]
print('PPG pencere: enerji =', np.sum(win**2), '  güç =', np.mean(win**2))
```
